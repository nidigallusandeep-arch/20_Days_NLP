# What are N-grams?
- N-grams are continuous sequences of N words in a sentence

# Why do we use N-grams?
- To understand word combinations.
- To capture some context between words.
- To improve text classification.
- To identify phrases in tex

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
```

```python
data = {
    'text': [
        'I love Python',
        'I love Machine Learning',
        'Python is easy',
        'Machine Learning is interesting',
        'I love Data Science',
        'Python is powerful',
        'Data Science is amazing',
        'I enjoy Machine Learning',
        'Python and Data Science',
        'I love Artificial Intelligence'
]}
```

```python
df = pd.DataFrame(data)
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.tail()
```

# Unigram using CountVectorizer

```python
unigram = CountVectorizer(ngram_range=(1, 1))
X_uni = unigram.fit_transform(df['text'])
print(unigram.get_feature_names_out())
```

```python
bigram = CountVectorizer(ngram_range=(2, 2))
X_bi = bigram.fit_transform(df['text'])
print(bigram.get_feature_names_out())
```

```python
trigram = CountVectorizer(ngram_range=(3, 3))
X_tri = trigram.fit_transform(df['text'])
print(trigram.get_feature_names_out())
```

```python
ngram = CountVectorizer(ngram_range=(1, 2))
X_ngram = ngram.fit_transform(df['text'])
print(ngram.get_feature_names_out())
```

```python
ngram_df = pd.DataFrame(X_ngram.toarray(),columns=ngram.get_feature_names_out())
ngram_df.head()
```

```python
final_df = pd.concat([df, ngram_df], axis=1)
final_df.head()
```

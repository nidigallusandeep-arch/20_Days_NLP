# What is Stemming
- Stemming is the process of removing prefixes or suffixes from words to get their root form.

# Example:
- playing  → play
- played   → play
- playing  → play

```python
import pandas as pd
import nltk
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
data = {
    'words': [
        'playing',
        'played',
        'running',
        'studies',
        'cars',
        'better',
        'eating',
        'writing',
        'happily',
        'children'
    ]
}

df = pd.DataFrame(data)

df
```

# Apply Stemming

```python
stemmer = PorterStemmer()
df['stemming'] = df['words'].apply(lambda x: stemmer.stem(x))
```

```python
df
```

- PorterStemmer() creates a stemming object.
- stem() converts a word into its stem.
- apply() applies the function to every row.

# Apply Lemmatization

```python
lemmatizer = WordNetLemmatizer()
df['lemmatization'] = df['words'].apply(lambda x: lemmatizer.lemmatize(x))
df
```

# lemmatization with POS

```python
words = ['playing', 'running', 'eating', 'studies']
for word in words:
    print(word, "→", lemmatizer.lemmatize(word, pos='v'))
```

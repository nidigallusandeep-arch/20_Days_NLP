# What is Word2Vec
- Word2Vec is a word embedding technique in NLP that converts words into numerical vectors.

# Why do we use Word2Vec?
1. To convert words into numerical vectors.
2. To understand word meanings.
3. To find similar words.
4. To capture relationships between words.
5. To improve NLP model performance.

```python
from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
sentences = [
    "I love machine learning",
    "Machine learning is very interesting",
    "I love artificial intelligence",
    "Artificial intelligence is the future",
    "Deep learning is part of machine learning",
    "Python is used for machine learning",
    "I love Python programming"
]
```

```python
tokenized_sentences = [
    word_tokenize(sentence.lower())
    for sentence in sentences]
print(tokenized_sentences)
```

```python
model=Word2Vec(
    sentences=tokenized_sentences,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4,
    sg=0
)
```

```python
vector=model.wv["machine"]
vector
```

```python
similar_words = model.wv.most_similar('machine', topn=3)
print(similar_words)
```

```python
print(model.wv.index_to_key)
```

```python
similarity = model.wv.similarity('machine', 'learning')
print(similarity)
```

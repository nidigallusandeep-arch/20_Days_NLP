# What is CountVectorizer?
- CountVectorizer is a feature extraction technique that converts text into numerical values by counting how many times each word appears.

# Why do we use CountVectorizer?
- Machine Learning models understand numbers, not raw text.
- It converts text into numerical data.
- It helps us prepare text for classification tasks.

```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
```

```python
data = {
    'text': [
        'I love Python',
        'I love Machine Learning',
        'Python is easy',
        'Machine Learning is interesting',
        'I love Data Science',
        'Python is powerful',
        'Data Science is amazing',
        'I enjoy Machine Learning',
        'Python and Data Science',
        'I love Artificial Intelligence'
    ]}
df = pd.DataFrame(data)
```

```python
df.head()
```

# Create CountVectorizer Object

```python
cv=CountVectorizer()
```

# Apply CountVectorizer

```python
x=cv.fit_transform(df["text"])
```

- fit() – learns all unique words.
- transform() – converts text into numerical values.
- fit_transform() – performs both operations together.

# Get Vocabulary

```python
vacabulary=cv.get_feature_names_out()
vacabulary
```

- This displays all unique words in the dataset.

# Convert Matrix into DataFrame

```python
count_df = pd.DataFrame(x.toarray(),columns=cv.get_feature_names_out())
count_df.head()
```

# Combine Original Text and Numerical Data

```python
final_df = pd.concat([df, count_df], axis=1)
final_df.head()
```

# Important CountVectorizer Parameters
- A. binary=True
- Counts are converted into 0 or 1. If a word appears once or many times, its value is 1.

```python
cv_binary = CountVectorizer(binary=True)
X_binary = cv_binary.fit_transform(df['text'])
pd.DataFrame(X_binary.toarray(),columns=cv_binary.get_feature_names_out())
```

# ngram_range=(1,2)
- It extracts single words and two-word combinations.

```python
cv_ngram = CountVectorizer(ngram_range=(1,2))
X_ngram = cv_ngram.fit_transform(df['text'])
print(cv_ngram.get_feature_names_out())
```

# stop_words='english'
- Removes common English stopwords.

```python
cv_stop = CountVectorizer(stop_words='english')
X_stop = cv_stop.fit_transform(df['text'])
print(cv_stop.get_feature_names_out())
```

# max_features=5
- Selects a maximum of 5 features (words).

```python
cv_max = CountVectorizer(max_features=5)
X_max = cv_max.fit_transform(df['text'])
print(cv_max.get_feature_names_out())
```

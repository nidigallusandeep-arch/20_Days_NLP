# What is Tokenization
- Tokenization is the process of splitting text into smaller units called tokens.

# Types of Tokenization
1. Word Tokenization
2. Sentence Tokenization

```python
import pandas as pd
import nltk
from nltk.tokenize import word_tokenize,sent_tokenize
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
data = {
    'text': [
        "I love Python programming.",
        "Machine learning is interesting.",
        "NLP helps computers understand language.",
        "Deep learning is powerful.",
        "I am learning Natural Language Processing."]}

df = pd.DataFrame(data)
```

```python
df
```

# Word Tokenization

```python
df["word_tokens"]=df["text"].apply(word_tokenize)
df[["text","word_tokens"]]
```

- word_tokenize() splits text into words and punctuation tokens.
- Each word becomes a separate token

# Sentence Tokenization

```python
text=""""I love Python.
Machine learning is interesting.
NLP is useful.
"""
```

```python
sentence=sent_tokenize(text)
sentence
```

- sent_tokenize() splits a paragraph into individual sentences

# Apply tokenization to dataset

```python
df['sentence_tokens'] = df['text'].apply(sent_tokenize)
df.head()
```

# Count words

```python
df['word_count'] = df['word_tokens'].apply(len)
df[['text', 'word_count']]
```

- len() counts the number of tokens in each sentence.

# Count total tokens

```python
total_tokens = sum(df['word_count'])
print("Total Tokens:", total_tokens)
```

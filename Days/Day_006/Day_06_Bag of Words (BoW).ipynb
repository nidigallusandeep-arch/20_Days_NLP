# What is Bag of Words
- Bag of Words is an NLP technique that converts text into numerical vectors based on word frequency.

```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
```

```python
data = {
    'text': [
        'I love Python',
        'I love Machine Learning',
        'Python is easy',
        'Machine Learning is interesting',
        'I love Data Science',
        'Python is powerful',
        'Data Science is amazing',
        'I enjoy Machine Learning',
        'Python and Data Science',
        'I love Artificial Intelligence'
]}

df = pd.DataFrame(data)
```

```python
df.head()
```

```python
cv=CountVectorizer()
X=cv.fit_transform(df["text"])
```

- CountVectorizer() creates the BoW object.
- fit() learns unique words from the dataset.
- transform() converts text into numerical counts.
- fit_transform() performs both operations.

```python
print(cv.get_feature_names_out())
```

- This displays all unique words in the dataset.

# Convert into DataFrame

```python
bow_df = pd.DataFrame(X.toarray(),columns=cv.get_feature_names_out())
bow_df
```

# Combine original text and BoW

```python
final_df = pd.concat([df, bow_df],axis=1)
final_df.head()
```

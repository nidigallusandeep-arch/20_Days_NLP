# What is tokenization?
Tokenization is the process of splitting text into smaller units called tokens.

# What is text preprocessing?
Text preprocessing is the process of cleaning and preparing text data before applying machine learning algorithms.

```python
import pandas as pd
import numpy as np
```

```python
data = {
    "text": [
        "I love Python",
        "Machine learning is interesting",
        "NLP is very useful",
        "I like artificial intelligence",
        "Python is easy to learn",
        "I love machine learning",
        "NLP helps computers understand text",
        "Deep learning is powerful",
        "I enjoy data science",
        "Artificial intelligence is amazing"]}
df = pd.DataFrame(data)
df.head()
```

# Convert text into lowercase

```python
df["clean_test"]=df["text"].str.lower()
```

```python
df.head()
```

# Tokenization

```python
df["tokens"]=df["clean_test"].apply(lambda x: x.split())
df.head()
```

- Tokenization is the process of splitting text into smaller units called tokens.

# What is Text Preprocessing
- Text preprocessing is the process of cleaning raw text data before giving it to a Machine Learning or NLP model.

```python
import pandas as pd
import re # re: Regular expressions for text cleaning.
```

```python
data = {
    "text": [
        "I LOVE Python!!!",
        "Machine Learning is 100% useful.",
        "NLP is amazing @2026",
        "Python@@ is very easy!!!",
        "I am learning NLP #AI",
        "Deep Learning costs 500 dollars.",
        "Hello!!! Welcome to NLP.",
        "Python is powerful $$$",
        "I love Data Science 123",
        "AI & ML are interesting."]}
df = pd.DataFrame(data)
```

# Apply Text Preprocessing
- Convert lowercase

```python
df["lowercase"]=df["text"].str.lower()
df[['text', 'lowercase']]
```

# Remove punctuation

```python
df['no_punctuation'] = df['lowercase'].apply(lambda x: re.sub(r'[^\w\s]', '', x))
df[['lowercase', 'no_punctuation']]
```

# Remove numbers

```python
df['no_numbers'] = df['no_punctuation'].apply(lambda x: re.sub(r'\d+', '', x))
df[['no_punctuation', 'no_numbers']]
```

# Remove special characters

```python
df['clean_text'] = df['no_numbers'].apply(lambda x: re.sub(r'[^a-zA-Z\s]', '', x))
df[['no_numbers', 'clean_text']]
```

# What are Stopwords?
- Stopwords are common words that usually carry less useful information for certain NLP tasks.
- Examples:
is, am, are, the, a, an, in, on, of, to

```python
import pandas as pd
import nltk
```

```python
data = {
    'text': [
        'I am learning Python',
        'Machine learning is very interesting',
        'I love natural language processing',
        'Python is easy to learn',
        'Deep learning is a part of machine learning',
        'I am working on NLP projects',
        'The data science field is amazing',
        'We are learning artificial intelligence',
        'This is a simple NLP example',
        'I like programming in Python'
    ]
}

df = pd.DataFrame(data)
```

```python
df.head()
```

# Apply tokenization

```python
df["tokens"]=df["text"].apply(lambda x:x.lower().split())
```

```python
df.head()
```

# Remove stopwords

```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))
```

```python
def remove_stopwords(text):
    words = text.lower().split()
    filtered_words = [
        word for word in words
        if word not in stop_words  ]
    return filtered_words
df['filtered_tokens'] = df['text'].apply(remove_stopwords)
```

```python
df.head()
```

# Convert tokens back into sentences

```python
df['clean_text'] = df['filtered_tokens'].apply(lambda x: ' '.join(x))
df[['text', 'clean_text']]
```

# Count words before and after

```python
df['before_count'] = df['text'].apply(lambda x: len(x.split()))
df['after_count'] = df['clean_text'].apply(lambda x: len(x.split()))
df[['text', 'before_count', 'after_count']]
```

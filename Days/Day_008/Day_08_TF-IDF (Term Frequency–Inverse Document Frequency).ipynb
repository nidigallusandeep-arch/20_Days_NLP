# What is TF-IDF?
- TF-IDF is a technique used to identify how important a word is in a document compared to a collection of documents.

# Why do we use TF-IDF?
- To find important words in text.
- To convert text into numerical features.
- To reduce the importance of common words.
- To prepare text for Machine Learning models.

```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
data = {
    'text': [
        'I love Python',
        'I love Machine Learning',
        'Python is easy',
        'Machine Learning is interesting',
        'I love Data Science',
        'Python is powerful',
        'Data Science is amazing',
        'I enjoy Machine Learning',
        'Python and Data Science',
        'I love Artificial Intelligence'
    ]}

df = pd.DataFrame(data)
```

```python
df.head()
```

```python
tfidf=TfidfVectorizer()
```

```python
x=tfidf.fit_transform(df["text"])
```

```python
features = tfidf.get_feature_names_out()
```

# Convert into DataFrame

```python
tfidf_df = pd.DataFrame(x.toarray(),columns=features)
tfidf_df.head()
```

# Combine Original Text and TF-IDF Values

```python
final_df = pd.concat([df, tfidf_df], axis=1)
final_df.head()
```

```python
print("Shape:", x.shape)
```

# Important TF-IDF Parameters
- A. max_features=5
- Selects a maximum of 5 features.

```python
tfidf_max = TfidfVectorizer(max_features=5)
X_max = tfidf_max.fit_transform(df['text'])
print(tfidf_max.get_feature_names_out())
```

# stop_words='english'
- Removes common English stopwords

```python
tfidf_stop = TfidfVectorizer(stop_words='english')
X_stop = tfidf_stop.fit_transform(df['text'])
print(tfidf_stop.get_feature_names_out())
```

# ngram_range=(1,2)
- Extracts single words and two-word combinations.

```python
tfidf_ngram = TfidfVectorizer(ngram_range=(1,2))
X_ngram = tfidf_ngram.fit_transform(df['text'])
print(tfidf_ngram.get_feature_names_out())
```

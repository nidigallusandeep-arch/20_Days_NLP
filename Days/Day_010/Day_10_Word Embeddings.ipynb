# What are Word Embeddings
- Word Embeddings are a technique used to represent words as dense numerical vectors. Words with similar meanings can have similar vector representations.

# Why do we use Word Embeddings?
- To convert words into numerical vectors.
- To capture semantic meaning.
- To identify relationships between words.
- To use text data in Deep Learning models.

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.layers import Embedding
```

```python
sentences = [
    "I love Python",
    "I love Machine Learning",
    "Python is easy",
    "Machine Learning is interesting",
    "I love Data Science",
    "Data Science is amazing"
]
```

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)
word_index = tokenizer.word_index
print(word_index)
```

```python
sequences = tokenizer.texts_to_sequences(sentences)
print(sequences)
```

```python
sequences = tokenizer.texts_to_sequences(sentences)
print(sequences)
```

```python
padded_sequences = pad_sequences(
    sequences,
    padding='post')
print(padded_sequences)
```

```python
vocab_size = len(word_index) + 1
embedding_dim = 5
embedding_layer = Embedding(
    input_dim=vocab_size,
    output_dim=embedding_dim)
```

```python
input_data = tf.constant(padded_sequences)
embedding_output = embedding_layer(input_data)
print(embedding_output.shape)
```

```python
word = "python"
word_id = word_index[word]
word_vector = embedding_layer(
    tf.constant([word_id]))
print("Word:", word)
print("Vector:", word_vector.numpy())
```
